```python
def square(number):
    return number * number

square(4)
```

## lambda arguments : expression

```python
# lambda function

square = lambda number : number * number

square(2)
```

```python
def is_adult(age):
    return age >= 18
is_adult(20)
```

```python
is_adult = lambda age : age>= 18

is_adult(18)
```

```python
x = {"queston" : "What is 2 + 3?"}
```

```python
def check_queston(x):
    for ch in x["queston"]:
        if ch.isdigit():
            return True
        else:
            return False
```

```python
# Smaller version
def check_queston(x):
    return any(ch.isdigit() for ch in x["queston"])
```

```python
check_queston(x)
```

```python
s = lambda x : any(ch.isdigit() for ch in x["queston"])
s(x)
```
